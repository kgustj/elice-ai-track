```python
def finite_generator():
    count = 0
    for _ in range(10):
        count += 1
        yield count
# next 한번마다 yield까지의 과정이 한번 실행 됨

gen = finite_generator()
next(gen)
next(gen)
next(gen)
```

```python
def finite_generator():
    yield 1
    yield 2
    yield 3
# next 한번마다 yield까지의 과정이 한번 실행 됨

gen = finite_generator()
print(next(gen))
print(next(gen))
print(next(gen))
```

```python
def f():
    count = 0
    while count < 3:
        count += 1
        yield count
        print(11)

# next 한번마다 yield까지의 과정이 한번 실행 됨

gen = f()
print(next(gen))
print("-" * 20)
print(next(gen))
print("-" * 20)
print(next(gen))
```
