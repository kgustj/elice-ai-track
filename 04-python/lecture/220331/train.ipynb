```python
def even_filter(nums):
    for num in nums:
        if num % 2 == 0:
            yield num

nums = [i for i in range(10)]
iter = even_filter(nums)

print(next(iter))
print(next(iter))
print(next(iter))
```

```python
space = [[0]*10 for _ in range(10)]
for i in space:
    print(i) 
```

```python
list = [[1, 4, 6], [2, 4, 6], [10, 4, 9], [-1, 3, 8]]

answer = max(map(lambda x: max(x), list))

print(answer)
```

```python
def main():
    
    numbers = [1,2,3,4,5,6]
    
    # 아래의 코드를 완성하세요.
    numbers_squared = list(map(lambda x: x ** 2, numbers))
    print(numbers_squared)

    # 아래의 코드를 완성하세요.
    numbers_even = list(filter(lambda x: x % 2 == 0, numbers))
    print(numbers_even)


if __name__ == "__main__":
    main()
```
