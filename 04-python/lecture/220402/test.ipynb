```python
def solution(string):
    bracket = 0

    for i in range(len(string)):
        if string[i] == "(":
            bracket += 1
        if string[i] == ")":
            bracket -= 1
        if bracket < 0:
            return "NO"

    if bracket == 0:
        return "YES"
    else:
        return "NO"


if __name__ == "__main__":
    print(solution(input()))
```
