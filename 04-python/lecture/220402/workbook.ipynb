```python
# 누가 기침 소리를 내었는가

# 리스트를 이용해 문제를 해결할 수 있습니다.
# n, c = map(int, input().split())
n, c = map(int, "2 20".split())
b = [0 for _ in range(c)]
m = [4, 6]
# for i in range(n):
#     m.append(int(input()))

# 길이가 c인 b 배열에서 기침을 하는 사람의 배수에 대한 인덱스의 값을 1로 설정합니다.
for i in m:
    for j in range(i-1, c, i):
        b[j] = 1

print(sum(b))
```

```python
# 초콜릿

# 리스트와 집합을 이용해 문제를 해결할 수 있습니다.
word = input()

slist = []
lenth = len(word)
# 입력된 문자열을 인덱싱하여 원소들에 대한 부분 집합을 구해 배열에 넣습니다.
for i in range(lenth,0,-1):
    for j in range(i):
        count = lenth - i + 1
        subset = word[j:j+count]
        slist.append(subset)
# 배열에서 중복된 문자는 제거합니다.
slist = list(set(slist))

print(len(slist)-1)
```

```python
# 차원 도망자

# 리스트와 튜플을 이용해 문제를 해결할 수 있습니다.
def compress(S):
    S = sorted(L)
    M = {}
    for i in range(len(S)): M[S[i]] = i
    return M

n = int(input())
m = int(input())
L = list(map(int,input().split()))
MAP = compress(L)
sign = input()

dim = [0] * m
trace = {tuple(dim)}
for i in range(m):
    # 방문 위치를 계산합니다.
    dim[MAP[L[i]]]+= 1 if sign[i] == "+" else -1
    T = tuple(dim)
    # 이미 방문한 곳이라면 0을 반환합니다.
    if T in trace: print(0); break
    trace.add(T)
else: print(1)
```

```python
# 치타의 동선

# 리스트를 이용해 문제를 해결할 수 있습니다.
import sys
# Counter는 카운팅을 쉽게 할 수 있게 돕는 라이브러리입니다.
# https://docs.python.org/ko/3/library/collections.html#collections.Counter
from collections import Counter

alpha = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
adict = Counter(alpha)
# sen = input().strip()
sen = "ABCCABDDEEFFGGHHIIJJKKLLMMNNOOPPQQRRSSTTUUVVWWXXYYZZ"
stack = []
count = 0
for i in sen:
    adict.update(i)
    if adict.get(i) == 3:
        # 알파벳 겹치게 되는 경우가 몇 개인지를 계산하고, 계산된 것은 제거합니다.
        count += len(stack) - stack.index(i) - 1
        stack.remove(i)
    else:
        # 리스트의 마지막에 알파벳을 넣어 겹치는 경우를 확인합니다.
        stack.append(i)
print(count)
```
