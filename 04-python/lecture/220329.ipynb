```python
class Person:
    def __init__(self, fname, lname):
        self.first = fname
        self.last = lname
    
    def printname(self):
        print(self.first, self.last)

x = Person("John", "Doe")
x.printname()

class Student(Person):
    def printname(self):
        # super().printname()
        print("S", self.first, self.last)

x = Student("Mike", "Olsen")
x.printname()
```
