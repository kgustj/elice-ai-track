```python
import numpy as np
print(np.__version__)
```

```python
# 제곱을 한 값을 list에 넣고, array로 만들어줌
%timeit np.array([x ** 2 for x in range(1000)])
# 0 ~ 999를 list에 넣고, array로 만든 뒤 제곱
%timeit np.array([x for x in range(1000)]) ** 2
# arange 함수를 통해 0 ~ 9로 이루어진 array를 만들고, 제곱
%timeit np.arange(1000) ** 2
```

```python
# zero matrix 만들기 (1차원 배열: vector, 2차원 배열: matrix)
zero_array_2d = np.zeros((2, 2))
print(zero_array_2d)
print(zero_array_2d.shape)
```

```python
# 9로만 구성된 ndarray 만들기 (float type)
one_array = np.ones((2, 2)) * 9
print(one_array)
```

```python
# 대각선이 1로 구성된 ndarray 만들기
eye_array = np.eye(2, 2)
print(eye_array)
```

```python
# 랜덤한 숫자로 이루어진 array 만들기
random_array = np.random.random((2, 2))
print(random_array)
```

```python
# arange 범위의 array 만들기
print(np.arange(3))
print(np.arange(3, 7))
print(np.arange(3, 7, 2))
print(np.arange(3, 8, 2))
```

```python
# zero_array_2d = [[0. 0.]
#                  [0. 0.]]
zero_array_1d = zero_array_2d.reshape(4)
print(zero_array_1d)
print(zero_array_1d.shape)
```

```python
array_1d = np.arange(8)
print(array_1d)

array_2d = array_1d.reshape((2, 4))
print(array_2d)

array_3d = array_1d.reshape((2, 2, 2))
print(array_3d)
```
